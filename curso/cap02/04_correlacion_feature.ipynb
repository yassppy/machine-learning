## ¿Qué es el Standard Correlation Coefficent?

El coeficiente de correlación es una medida estadística que nos da dos pistas clave sobre la relación entre dos variables numéricas:
1. La fuerza: Qué tan "apegadas" o conectadas están (si una cambia, ¿la otra cambia de inmediato o le da igual?).

2. La dirección: Si se mueven hacia el mismo lado o en sentidos opuestos.

⚠️ Regla de oro: Este coeficiente solo mide relaciones lineales (Líneas rectas). Si la relación tiene forma de curva o patrones extraños, este número no funcionará bien. Sin correlación puede ser que exista patrones internas que la correlación de pearson no puede estar en la capacidad de detectar y se puebra otras formas.

$$r = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2 \cdot \sum_{i=1}^{n} (y_i - \bar{y})^2}}$$

- $r$: Es el resultado final (el número entre -1 y 1 que viste en la infografía).
- $x_i$ e $y_i$: Son los valores de las dos variables en una fila específica (por ejemplo, el ingreso medio y el precio de la casa del distrito número 5).
- $\bar{x}$ e $\bar{y}$: Son los promedios (medias) de toda la columna $X$ y toda la columna $Y$.
- El numerador (arriba) $\sum (x_i - \bar{x})(y_i - \bar{y})$: Mide la Covarianza. Multiplica qué tanto se aleja cada dato de su promedio. Si ambas variables suelen estar por encima de su promedio al mismo tiempo, este número crece en positivo.
- El denominador (abajo) $\sqrt{\sum (x_i - \bar{x})^2 \cdot \sum (y_i - \bar{y})^2}$: Es la multiplicación de las desviaciones estándar. Sirve como un "normalizador" para obligar a que el resultado final del rango no se dispare al infinito y quede atrapado estrictamente entre -1 y 1.

## Formula de correlación de Pearson

Mide que tan alineadas están 2 variables, si una sube la otra sube y si baja lo mismo.

Interpretación del rango

- r = 1 → Correlación positiva perfecta
- r = -1 → Correlación negativa perfecta
- r = 0 → Sin correlación lineal
- 0.7 < r < 1 → Correlación positiva fuerte
- -1 < r < -0.7 → Correlación negativa fuerte

## Covarianza

- La covarianza sí te dice la dirección (si es una línea positiva, negativa o neutra) observando como varian juntos esas variables.
- El cov(X, Y) se va a representar como S(X,Y) 
$$s_{xy} = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{n - 1}$$

> Punto débil de la covarianza y por que se invento la c pearson, es imposible saber si una relación es fuerte o débil usando solo la covarianza porque el resultado cambia por completo según la unidad de medida que elijas.
> - Ejemplo: En Kilómetros: Covarianza = $+80$ y En Metros: Covarianza = $+80,000$

## Normalización

- Al normalizar la covarianza, Pearson elimina las unidades de medida y nos permite medir la fuerza y dirección de la relación en un rango fijo de -1 a 1.
- Se llama "standard" porque está estandarizado. La escala siempre se cancela → r siempre entre -1 y 1

- Limitaciones Críticas (Solo relación lineal):
  - No detecta relaciones complejas en forma de curvas, parábolas (U), círculos o letras
  - Un valor de $r = 0$ solo significa que no hay una relación lineal, pero las variables podrían estar fuertemente relacionadas de forma no lineal (curva)
  - Que $r = 1$ o $r = -1$ solo significa que los puntos caen perfectamente sobre una recta; no te dice si la pendiente de esa recta es empinada o plana
  - Regla de oro: Nunca te fíes solo del número $r$. Siempre visualiza el gráfico de dispersión (Scatter Plot) antes de sacar conclusiones.

Tomar variables que ya tenemos para modificarlas en información más útil.
